```python

import random

subjects = ["AI", "DBMS", "Java", "OS"]
difficulty = [5, 3, 4, 2]
H, N, iterations = 8, 10, 100

D = sum(difficulty)
T = [d / D * H for d in difficulty]

# Fitness function
def fitness(s):
    return sum((s[i] - T[i])**2 for i in range(4))

# Initialize cell
def create_cell():
    s = [0]*4
    for _ in range(H):
        s[random.randrange(4)] += 1
    return s

# Mutation
def mutate(s):
    a, b = random.sample(range(4), 2)
    new = s.copy()
    if new[a] > 0:
        new[a] -= 1
        new[b] += 1
    return new

# Initialize population
cells = [create_cell() for _ in range(N)]
best = min(cells, key=fitness)

# PCA iterations
for _ in range(iterations):
    new_cells = []

    for i in range(N):
        neighbors = [cells[(i-1)%N], cells[(i+1)%N]]
        current = min([cells[i]] + neighbors, key=fitness)

        candidate = mutate(current)
        new_cells.append(min([current, candidate], key=fitness))

    cells = new_cells
    best = min([best] + cells, key=fitness)

# Output
print("Study Time Allocation")
for s, h in zip(subjects, best):
    print(s, ":", h, "hours")

print("Total Hours:", sum(best))
print("Fitness:", round(fitness(best), 4))
```
