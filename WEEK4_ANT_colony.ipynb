```python

import random

n = int(input("Enter number of cities: "))
d = []

for i in range(n):
    d.append(list(map(int, input().split())))

p = [[1]*n for _ in range(n)]
best = 9999

for _ in range(10):
    route = [0] + random.sample(list(range(1, n)), n-1) + [0]
    dist = sum(d[route[i]][route[i+1]] for i in range(n))

    if dist < best:
        best = dist
        ans = route

    for i in range(n):
        a, b = route[i], route[i+1]
        p[a][b] += 1/dist

    for i in range(n):
        for j in range(n):
            p[i][j] *= 0.9

print("Best Route:", ans)
print("Shortest Distance:", best)

```
